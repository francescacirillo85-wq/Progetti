```python
# Simulating slot machine game with 50% win probability and €100 budget

import random
import matplotlib.pyplot as plt
import os

# Impostazioni iniziali
budget_iniziale = 100
budget = budget_iniziale
costo_giocata = 1
probabilita_vittoria = 0.5
vincita_per_vittoria = 2  # Guadagno netto per ogni vittoria

# Variabili per tracciare l'andamento
giocate = 0
vittorie = 0
andamento_budget = [budget]

# Definizione del budget obiettivo
target_budget = 101

# Simulazione del gioco
while budget >= costo_giocata and budget < target_budget:
    giocate += 1
    budget -= costo_giocata
    if random.random() < probabilita_vittoria:
        vittorie += 1
        budget += vincita_per_vittoria
    andamento_budget.append(budget)

# Creazione del grafico
plt.style.use('seaborn-v0_8')
plt.figure(figsize=(10, 6))
plt.plot(andamento_budget, label='Budget (€)', color='blue')
plt.axhline(y=budget_iniziale, color='gray', linestyle='--', label='Budget iniziale')
plt.axhline(y=target_budget, color='green', linestyle=':', label='Budget obiettivo (>100 €)')
plt.title('Andamento del Budget nel Gioco della Slot Machine')
plt.xlabel('Numero di giocate')
plt.ylabel('Budget (€)')
plt.legend()
plt.grid(True)

# Salvataggio del grafico
output_path = "/mnt/data/andamento_slot_machine.png"
os.makedirs(os.path.dirname(output_path), exist_ok=True) # Ensure the directory exists
plt.savefig(output_path)

# Risultati
print(f"Giocate effettuate: {giocate}")
print(f"Numero di vittorie: {vittorie}")
print(f"Budget finale: {budget} €")

if budget >= target_budget:
    print(f"Complimenti! Il budget ha raggiunto l'obiettivo di {target_budget} € dopo {giocate} giocate.")
elif budget < costo_giocata:
    print(f"Il budget si è esaurito prima di raggiungere l'obiettivo di {target_budget} €.")

print("Grafico salvato come 'andamento_slot_machine.png'")
```
